```python
import subprocess
import os
import sys
import re
import numpy as np
from scipy.optimize import minimize
```

```python
def rosen(x):
    """The Rosenbrock function"""
    return sum(100.0*(x[1:]-x[:-1]**2.0)**2.0 + (1-x[:-1])**2.0)
```

```python
x0 = np.array([1.3, 0.7, 0.8, 1.9, 1.2])
```

```python
res = minimize(rosen, x0, method='nelder-mead',
                options={'xtol': 1e-8, 'disp': True})
```

```python
print(res.x)
```

```python
bindir='./bin'
data_path='./data'
output_directory='./'
seqfile='.dat'
```

```python
bindir=os.path.realpath(bindir)
data_path = os.path.realpath(data_path)

if not os.path.isdir(output_directory):
    os.makedirs(output_directory)
```

```python
def getEQ(x_,model=model):
    EQ=' " '
    for i in range(len(x_)):
        if len(EQ)>0:
            EQ=EQ+' + '
        EQ=EQ+'( '+"%.9f" % x_[i]+' * '+model[i]+' ) '
    EQ=EQ+'"'
    return EQ

def loss(x_):
    EQ=getEQ(x_,model)
    PSOLVE=os.path.join(bindir,'psolve')+' -D "" -o '+\
                                os.path.join(output_directory,'machine.cfg')+' -x '+EQ
    LLK=os.path.join(bindir,'llk')+' -f '+os.path.join(output_directory,'machine.cfg')+\
                                    ' -T symbolic -s '+os.path.join(data_path,seqfile)
    ret=subprocess.check_output(PSOLVE, shell=True)
    ret=subprocess.check_output(LLK, shell=True)
    #print PSOLVE
    return float(re.findall("\d+\.\d+", ret)[0])

```

```python
model=['M2.cfg','M4.cfg','S2.cfg','T3.cfg']
model=['T3.cfg']
```

```python
x0_=[ 1.0]
```

```python
loss(x0_)
```

```python
x0=[0.1]
res = minimize(loss, x0, method='nelder-mead',
                options={'xtol': 1e-10, 'disp': True})
```

```python
loss(res.x)
```
