```python
import pandas as pd
import numpy as np
from glob import glob
from time import time

import numpy.linalg as LA
```

## Test accuracy

```python
result = []
index = []

time0 = time()
for pfsa in sorted(glob('*pfsa')):
    prefix = pfsa.rstrip('.pfsa')
    print(prefix)
    
    out = prefix + '.out'
    ! ./llk -s testfile.dat -f {pfsa} > {out}
    res = np.genfromtxt(out)
    
    index.append(prefix)
    result.append(res)
time1 = time()

time_elapsed = (time1 - time0) / 60.
print('time elapsed = {} min.s'.format(time_elapsed))
    
df = pd.DataFrame(index=index, data=result)
df.to_csv('llk_result.csv')
```

```python
A = pd.read_csv('llk_result.csv', index_col=0).values
B = pd.read_csv('llk_result_2.csv', index_col=0).values
D = A - B
```

```python
from scipy.spatial import distance
for a, b, d in zip(A, B, D):
    na = LA.norm(a)
    nb = LA.norm(b)
    nd = LA.norm(d)
    print(distance.cosine(a, b))
    print('na={}, nb={}, nd={}\n'.format(na, nb, nd))
```
