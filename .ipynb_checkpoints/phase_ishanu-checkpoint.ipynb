# In this notebook, we get the non-trivial solutions for $\left\{(\alpha, \beta):\ L_{\beta G}\left(T(\alpha G, l)\right) - \mathcal{H}(\beta G) = 0\right\}$

We find nontrivial $f_{G, T}(\alpha)$ such that $$L_{f(\alpha) G}\left(T(\alpha G, l)\right) - \mathcal{H}(f(\alpha) G) = 0$$

We note that $f(\alpha) = 0$ is a trivial solution.
- The way we find the non-trivial part of the solution is empirical;
- and the approximated curve can be very noisy.
- But from the approximated curve, we can see that true curve is rather smooth.

We process the approximated curve and fit a polynomial as a better approximation to $f(\alpha)$.

```python
import numpy as np
import seaborn as sns
sns.set(color_codes=True)
import matplotlib.pyplot as plt
from scipy.signal import savgol_filter
```

```python
from PFSA_basic import PFSA
import PFSA_generation as gen
import PFSA_phase as phase
import PFSA_linearSpace as ls
import digraph
import util
import Sequence_processing as sp
```

```python
def run(G, H, length):
    input_seq = G.generate_sequence(length)
    proj_sd = phase.get_proj_stationary(G, H.Connex)
    output_seq = phase.run(input_seq, H.Connex, H.Pitilde, proj_sd)
    return output_seq
```

```python
def smooth(y, box_pts):
    box = np.ones(box_pts)/box_pts
    y_smooth = np.convolve(y, box, mode='same')
    return y_smooth


def remove_outliers(x, y, box_pts, z_value_cutoff):
    y_smooth = smooth(y, box_pts)
    
    diff = y_smooth - y
    mean = np.mean(diff)
    std = np.std(diff)
    
    outlier_idx = []
    for i, d in enumerate(diff):
        z_value = (d - mean) / std
        if abs(z_value) > 2:
            outlier_idx.append(i)
    
    cleaned_x = np.delete(x, outlier_idx)
    cleaned_y = np.delete(y, outlier_idx)
    
    return cleaned_x, cleaned_y
```

```python
cfg_folder = '/home/yhuang10/D3M/zutil_/cfgfiles'
folder = '/home/yhuang10/D3M/zutil_/llk_matrices'
```

```python
def analyze_result(xpfsa, pfas, sanity_check=False):
    
    # Get data (has to be generated first by c++)
    llk_matrix_filename = '{}/llk_matrix-scaled-{}-{}'.format(folder, pfsa, xpfsa)
    llk_matrix = np.genfromtxt(llk_matrix_filename)
    er_filename = '{}/er-scaled-{}'.format(folder, pfsa)
    er = np.genfromtxt(er_filename)[1]
    X = llk_matrix[0]
    llk_matrix = llk_matrix[1:]
    
    # Get (alpha, beta) pair and process
    X_0 = []
    for i in range(len(X)):
        vec = abs(llk_matrix[i] - er)
        idx = np.argsort(vec)[:3]
        x = X[idx]
        idx_largest_dev = np.argmax(abs(x))
        X_0.append(x[idx_largest_dev])

    cleaned_x, cleaned_y = remove_outliers(X, X_0, 5, 2)
    f = np.poly1d(np.polyfit(cleaned_x, cleaned_y, 3))  ###################### We find the f here ########################
    
    # Plot
    plt.figure(figsize=(6, 6))
    plt.xlabel(r'$\alpha$', fontsize=16)
    plt.ylabel(r'$\beta$', fontsize=16, rotation=0)

    plt.plot(X, X_0, 'g-', label='original')
    plt.plot(cleaned_x, cleaned_y, '-', label='outlier removed')
    plt.plot(X, f(X), 'r--', label=r'poly fitted $f(\alpha)$')
    plt.title('Non-trival Solution, Cleaned, and Fitted', fontsize=15)

    plt.legend(fontsize=15)
    plt.show()
    
    
    # Check whether f(alpha) works well in solving the equation
    if sanity_check:
        length = 5000

        Connex, Pitilde = util.read_C_result('{}/{}.cfg'.format(cfg_folder, pfsa))
        Eta, Kai = util.read_C_result('{}/{}.cfg'.format(cfg_folder, xpfsa))

        G = PFSA(Connex=Connex, Pitilde=Pitilde)
        H = PFSA(Connex=Eta, Pitilde=Kai)

        in_seq = G.generate_sequence(5000)
        out_seq = phase.run(in_seq, Eta, Kai, phase.get_proj_stationary(G, Eta))

        Diff = []
        for x in X:
            y = f(x)
            yG = ls.scalarProd(y, G)
            xG = ls.scalarProd(x, G)

            seq = run(xG, H, length)
            diff = yG.get_log_likelihood(seq) -  yG.get_entropy_rate()

            Diff.append(diff)
    
        plt.figure()
        plt.plot(X, np.abs(Diff), color='b')
        plt.axhline(max(np.abs(Diff)), color='r', label='max')
        plt.axhline(min(np.abs(Diff)), color='g', label='min')
        plt.axhline(np.mean(np.abs(Diff)), color='k', label='mean')
        plt.legend(fontsize=15)
        plt.xlabel(r'$\alpha$', fontsize=14)
        plt.ylabel(r'$L_{f(\alpha) G}\left(T(\alpha G, l)\right) - \mathcal{H}(f(\alpha) G)$', fontsize=14)
        plt.title(r'Check the Qualtity of $f(\alpha)$ by $\left|L_{f(\alpha) G}\left(T(\alpha G, l)\right) - \mathcal{H}(f(\alpha) G)\right|$')
        plt.show()
    
    
    return X, X_0, er, f
```

```python
xpfsa, pfsa = "XPFSA_2_3_2", "PFSA_M2_3"
Alpha_1, Beta_1, er_1, f_1 = analyze_result(xpfsa, pfsa, sanity_check=True)
```

```python
xpfsa, pfsa = "XPFSA_2_3_2", "PFSA_M2_0"
Alpha_2, Beta_2, er_2, f_2 = analyze_result(xpfsa, pfsa, sanity_check=True)
```
